```python
kpi_summary = {
    "Total Customers": df_clean["Customer ID"].nunique(),
    "Total Purchase Amount": df_clean["Purchase Amount (USD)"].sum(),
    "Average Purchase Amount": df_clean["Purchase Amount (USD)"].mean(),
    "Average Review Rating": df_clean["Review Rating"].mean(),
    "Average Previous Purchases": df_clean["Previous Purchases"].mean()
}

kpi_summary
```

```python
category_analysis = df_clean.groupby("Category").agg(
    Customers=("Customer ID", "nunique"),
    Total_Purchase_Amount=("Purchase Amount (USD)", "sum"),
    Average_Purchase_Amount=("Purchase Amount (USD)", "mean")
).sort_values("Total_Purchase_Amount", ascending=False)

category_analysis
```

```python
season_analysis = df_clean.groupby("Season").agg(
    Customers=("Customer ID", "nunique"),
    Total_Purchase_Amount=("Purchase Amount (USD)", "sum"),
    Average_Purchase_Amount=("Purchase Amount (USD)", "mean")
).sort_values("Total_Purchase_Amount", ascending=False)

season_analysis
```

```python
subscription_analysis = df_clean.groupby("Subscription Status").agg(
    Customers=("Customer ID", "nunique"),
    Total_Purchase_Amount=("Purchase Amount (USD)", "sum"),
    Average_Purchase_Amount=("Purchase Amount (USD)", "mean"),
    Average_Previous_Purchases=("Previous Purchases", "mean")
).sort_values("Total_Purchase_Amount", ascending=False)

subscription_analysis
```

```python
discount_analysis = df_clean.groupby("Discount Applied").agg(
    Customers=("Customer ID", "nunique"),
    Total_Purchase_Amount=("Purchase Amount (USD)", "sum"),
    Average_Purchase_Amount=("Purchase Amount (USD)", "mean"),
    Average_Previous_Purchases=("Previous Purchases", "mean")
).sort_values("Total_Purchase_Amount", ascending=False)

discount_analysis
```

```python
payment_analysis = df_clean.groupby("Payment Method").agg(
    Customers=("Customer ID", "nunique"),
    Total_Purchase_Amount=("Purchase Amount (USD)", "sum"),
    Average_Purchase_Amount=("Purchase Amount (USD)", "mean")
).sort_values("Total_Purchase_Amount", ascending=False)

payment_analysis
```

```python
frequency_analysis = df_clean.groupby("Frequency of Purchases").agg(
    Customers=("Customer ID", "nunique"),
    Total_Purchase_Amount=("Purchase Amount (USD)", "sum"),
    Average_Purchase_Amount=("Purchase Amount (USD)", "mean"),
    Average_Previous_Purchases=("Previous Purchases", "mean")
).sort_values("Customers", ascending=False)

frequency_analysis
```

```python
gender_analysis = df_clean.groupby("Gender").agg(
    Customers=("Customer ID", "nunique"),
    Total_Purchase_Amount=("Purchase Amount (USD)", "sum"),
    Average_Purchase_Amount=("Purchase Amount (USD)", "mean"),
    Average_Previous_Purchases=("Previous Purchases", "mean")
).sort_values("Customers", ascending=False)

gender_analysis
```

```python
age_analysis = df_clean.assign(
    Age_Group=pd.cut(
        df_clean["Age"],
        bins=[17, 24, 34, 44, 54, 64, 70],
        labels=["18-24", "25-34", "35-44", "45-54", "55-64", "65-70"]
    )
).groupby("Age_Group", observed=True).agg(
    Customers=("Customer ID", "nunique"),
    Total_Purchase_Amount=("Purchase Amount (USD)", "sum"),
    Average_Purchase_Amount=("Purchase Amount (USD)", "mean"),
    Average_Previous_Purchases=("Previous Purchases", "mean")
)

age_analysis
```

```python
category_gender_analysis = pd.pivot_table(
    df_clean,
    index="Category",
    columns="Gender",
    values="Customer ID",
    aggfunc="nunique",
    fill_value=0
)

category_gender_analysis
```

```python
category_gender_pct = pd.crosstab(
    df_clean["Gender"],
    df_clean["Category"],
    normalize="index"
) * 100

category_gender_pct.round(2)
```

```python
location_analysis = df_clean.groupby("Location").agg(
    Customers=("Customer ID", "nunique"),
    Total_Purchase_Amount=("Purchase Amount (USD)", "sum"),
    Average_Purchase_Amount=("Purchase Amount (USD)", "mean")
).sort_values("Total_Purchase_Amount", ascending=False)

location_analysis.head(10)
```

```python
shipping_analysis = df_clean.groupby("Shipping Type").agg(
    Customers=("Customer ID", "nunique"),
    Total_Purchase_Amount=("Purchase Amount (USD)", "sum"),
    Average_Purchase_Amount=("Purchase Amount (USD)", "mean")
).sort_values("Customers", ascending=False)

shipping_analysis
```

```python
rating_analysis = df_clean.groupby("Review Rating").agg(
    Customers=("Customer ID", "nunique"),
    Total_Purchase_Amount=("Purchase Amount (USD)", "sum"),
    Average_Purchase_Amount=("Purchase Amount (USD)", "mean")
).sort_index()

rating_analysis
```

```python
df_clean[["Review Rating", "Purchase Amount (USD)"]].corr()
```

```python
eda_summary = pd.DataFrame({
    "Metric": [
        "Total Customers",
        "Total Recorded Purchase Amount",
        "Average Purchase Amount",
        "Average Review Rating",
        "Average Previous Purchases",
        "Customers with Subscription",
        "Customers with Discount",
        "Missing Review Ratings",
        "Review Rating vs Purchase Correlation"
    ],
    "Value": [
        df_clean["Customer ID"].nunique(),
        df_clean["Purchase Amount (USD)"].sum(),
        round(df_clean["Purchase Amount (USD)"].mean(), 2),
        round(df_clean["Review Rating"].mean(), 2),
        round(df_clean["Previous Purchases"].mean(), 2),
        (df_clean["Subscription Status"] == "Yes").sum(),
        (df_clean["Discount Applied"] == "Yes").sum(),
        df_clean["Review Rating"].isnull().sum(),
        round(
            df_clean[["Review Rating", "Purchase Amount (USD)"]]
            .corr()
            .loc["Review Rating", "Purchase Amount (USD)"],
            4
        )
    ]
})

eda_summary
```
